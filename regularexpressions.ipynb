```python
import re
```

```python
#match(pat,string)
string = "The sky is blue and Ocean is blue"
res = re.match("[A-Z]" , string)
print(res)
print(res.span()) #it returns the matched positions
print(res.group())#it return matched pattern val
```

```python
string = "The sky is blue and Ocean is blue"
res = re.match("[A-Za-z]+" , string)
print(res)
print(res.span()) #it returns the matched positions
print(res.group())#it return matched pattern val
```

```python
#match(pat, string)
try:
    string = "The sky is blue and Ocean is blue"
    res = re.match("[z]+" , string)
    print(res)
    print(res.span()) #it returns the matched positions
    print(res.group())#it return matched pattern val
except Exception as e:
    print("There is no matched pattern:",e)
```

```python
#search(pat, string)
#search 5 length word
#match(pat, string)
import re
try:
    string = "The sky is blue and Ocean is blue"
    res = re.search("[a-zA-Z]{5}" , string)
    print(res)
    print(res.span()) #it returns the matched positions
    print(res.group())#it return matched pattern val
except Exception as e:
    print("There is no matched pattern:",e)
```

```python
#get the words from string which length is more than  3
#finadall(pat, string)
import re
try:
    string = "The sky is blue and Ocean is blue"
    res = re.findall("[a-zA-Z]{3,}" , string)
    print(res)
except Exception as e:
    print("There is no matched pattern:",e)
```

```python
import re
try:
    string = "The sky is blue and Oce456an is b123lue"
    res = re.findall("[a-zA-Z]+" , string)
    print(res)
except Exception as e:
    print("There is no matched pattern:",e)
```

```python
#validate mobile nuber
#All are digits,length of number is 10, starts with 6,7,8,9,Remaining are 0-9
import re
try:
    mobile =input()
    res = re.fullmatch("^[6-9][0-9]{9}", mobile)
    print(res)
    if res:
        print(f"{mobile} is valid")
    else:
        print(f"{mobile} is Invalid")
except Exception as e:
    print("There is no matched pattern:",e)
```

```python
#validate pan number
#it allow upper and digits,length is 10,first 5 are upper,next 4 are digitss and last
import re
try:
    PAN =input()
    res = re.fullmatch("^[A-Z]{5}[0-9]{4}[A-Z]{1}", PAN)
    print(res)
    if res:
        print(f"{PAN} is valid")
    else:
        print(f"{PAN} is Invalid")
except Exception as e:
    print("There is no matched pattern:",e)
```
