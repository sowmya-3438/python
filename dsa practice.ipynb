```python
s = input()

words = s.split()

freq = {}

for word in words:
    if word in freq:
        freq[word] += 1
    else:
        freq[word] = 1

max_count = 0
most_frequent = ""

for word in freq:
    if freq[word] > max_count:
        max_count = freq[word]
        most_frequent = word

print("Most frequent word:", most_frequent)
print("Frequency:", max_count)
```
