```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Housing.csv")
a = sns.countplot(x = 'guestroom',data= df)
for bars in a.containers:
    a.bar_label(bars)
```

```python
ax = sns.countplot(data = df, x = 'bedrooms', hue = 'bathrooms')
for bars in ax.containers:
    ax.bar_label(bars)
```
