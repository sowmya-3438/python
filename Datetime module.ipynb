```python
from datetime import datetime, timedelta
cur_timestamp = datetime.now()#to get current timestamp
print(cur_timestamp)
today = cur_timestamp.date()#it exteacts data from timestamp
cur_time = cur_timestamp.time()#it extract time
print(today)
print(cur_time)
print(today.year)
print(today.month)
print(today.day)
print(cur_time.hour ,cur_time.minute, cur_time.second)
print(today.strftime("%d-%m-%Y"))
print(cur_time.strftime("%I:%M:%p"))
```
